Frente a la pregunta "¿Por qué son 144 horas diferentes?", la respuesta es que este número corresponde a la cantidad de combinaciones posibles de tiempo bajo un formato de 24 horas cuando los datos se registran cada 10 minutos.

En un **día existen 24 horas posibles** y, **dentro de cada hora, pueden registrarse seis intervalos de 10 minutos**: 00, 10, 20, 30, 40 y 50. Por lo tanto, **el número total de combinaciones posibles se obtiene al multiplicar las 24 horas por los 6 intervalos de tiempo**:

**24 × 6 = 144**

En consecuencia, **el conjunto de datos puede contener como máximo 144 valores únicos de hora**, ya que no es posible generar más combinaciones bajo estas condiciones de medición.

8. Utiliza el dataframe de indicadores diarios para crear un nuevo dataframe `indicadores_mensuales`. Para ello:
- Crea una nueva columna `mes` a partir de la fecha que se encuentra en el índice.
- Haciendo uso de `groupby()`, calcula los promedios mensuales y conserva únicamente las columnas `humedad_relativa`, `temperatura_celsius` y `velocidad_viento` en el nuevo dataframe.

```python
# Crear columna 'mes' a partir del índice, agrupar, calcular promedios
# y conservar solo las 3 columnas solicitadas
indicadores_mensuales = indicadores_diarios.copy()
indicadores_diarios.index = pd.to_datetime(indicadores_diarios.index)
indicadores_mensuales['mes'] = indicadores_mensuales.index.month
# aplicar groupby por la nueva columna 'mes', calcular la media (.mean())
# y seleccionar únicamente las columnas requeridas
indicadores_mensuales = indicadores_mensuales.groupby('mes')[['humedad_relativa',
'temperatura_celsius', 'velocidad_viento']].mean().reset_index()
print("DataFrame 'indicadores_mensuales':")
indicadores_mensuales
```

9. Construye el mismo dataframe mensual, pero esta vez utilizando la función `pivot_table()`.
- Añade la columna `indice_calor` usando la fórmula de Thom (*Temperature - Humidity Index*, TDI):
> `temperatura_celsius + 0.33 x humedad_relativa - 0.7 x velocidad_viento - 4`

```python
# Crear columna 'mes' a partir del índice, agrupar, calcular promedios
# y conservar solo las 3 columnas solicitadas
indicadores_mensuales = indicadores_diarios.copy()
indicadores_diarios.index = pd.to_datetime(indicadores_diarios.index)
indicadores_mensuales['mes'] = indicadores_mensuales.index.month
# Construir con la funcion pivot_table y se calcula el promedio agrupado por la columna 'mes'
indicadores_mensuales = indicadores_mensuales.pivot_table(values=['humedad_relativa', 'temperatura_celsius',
'velocidad_viento'],index='mes',aggfunc='mean').reset_index()

#Se construye el calculo para agregar usando la fórmula de Thom (TDI)
# Fórmula: temperatura_celsius + 0.33 * humedad_relativa - 0.7 * velocidad_viento - 4
indicadores_mensuales['indice_calor'] = (indicadores_mensuales['temperatura_celsius']
+ 0.33 * indicadores_mensuales['humedad_relativa'] - 0.7 * indicadores_mensuales['velocidad_viento']- 4)

print("DataFrame 'indicadores_mensuales' con pivot_table e índice de calor:")
indicadores_mensuales
```

10. Recuerda que de las cuatro funciones estudiadas para manipular la estructura del dataframe:
- `melt/pivot` hacen las transformaciones de manera controlada, es decir puedes
definir por medio de sus parámetros qué variables quedarán como índices, columnas y valores.
- `stack/unstack` la conversión se aplica siempre sobre los niveles inferiores de index/columns.
- Convierte el dataframe del ejercicio anterior a formato largo usando: `melt()` y `stack()` y comenta las diferencias.

```python
#pd.melt() transforma un DataFrame de formato ancho a largo
# - id_vars: columnas que se mantienen como identificadores (no se “derriten”)
# - var_name: nombre de la nueva columna que contendrá los nombres de las columnas originales “derretidas”
# - value_name: nombre de la nueva columna que contendrá los valores correspondientes
indicadores_mensuales = pd.DataFrame(indicadores_mensuales)
indicadores_mensuales.reset_index(inplace=True)
indicadores_mensuales_long = indicadores_mensuales.melt(id_vars=['mes'], value_vars=['humedad_relativa', 'temperatura_celsius','velocidad_viento'],
                                                        var_name='variables climaticas', value_name='Value')
print("DataFrame 'indicadores_mensuales' en formato largo:")
indicadores_mensuales_long.head(20)
```

```python
## pd.stack() convierte columnas en filas creando un MultiIndex en las filas, transformando el dataFrame ancho en largo
indicadores_mensuales = indicadores_mensuales.stack()
indicadores_mensuales.head(20)
```
